### Подготовка и загрузка данных

```python
import os

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()

DATABASE_URL = os.getenv("DATABASE_URL")

engine = create_engine(DATABASE_URL)
```

```python
query = """
SELECT
    fo.order_id,
    dc.customer_unique_id,
    fo.order_purchase_timestamp,
    fo.order_status,
    fo.total_order_amount,
    fo.delivery_days,
    fo.days_late,
    fo.is_delayed,
    fo.review_score
FROM dwh.fact_orders fo
JOIN dwh.dim_customers dc
    ON fo.customer_id = dc.customer_id
"""

orders = pd.read_sql(query, engine)

orders.head()
orders.info()
orders.describe()
```

### Анализ повторных покупок

```python
customer_stats = (
    orders.groupby("customer_unique_id")
    .agg(
        orders_count=("order_id", "nunique")
    )
)

customer_stats["is_repeat"] = customer_stats["orders_count"] > 1

total_customers = len(customer_stats)
repeat_customers = customer_stats["is_repeat"].sum()
repeat_rate = repeat_customers / total_customers * 100

print(f"Всего клиентов: {total_customers}")
print(f"Повторных покупателей: {repeat_customers}")
print(f"Доля повторных покупателей: {repeat_rate:.2f}%")
```

```python
orders["purchase_date"] = orders["order_purchase_timestamp"].dt.date

customer_stats = (
    orders.groupby("customer_unique_id")
    .agg(
        orders_count=("purchase_date", "nunique")
    )
)

customer_stats["is_repeat"] = customer_stats["orders_count"] > 1

orders = orders.merge(
    customer_stats[["is_repeat"]],
    on="customer_unique_id",
    how="left"
)
```

```python
def delivery_group(x):
    if pd.isna(x):
        return "3. не доставлен"
    return "2. с задержкой" if x else "1. вовремя"


first = (
    orders.sort_values("order_purchase_timestamp")
    .drop_duplicates("customer_unique_id", keep="first")
    .copy()
)

first["delivery"] = first["is_delayed"].apply(delivery_group)

result = (
    first.groupby("delivery")["is_repeat"]
    .agg(
        customers="count",
        repeat_customers="sum"
    )
)

result["repeat_pct"] = (
    result["repeat_customers"] / result["customers"] * 100
).round(2)

result
```

### Анализ повторных покупок по периодам

```python
orders["purchase_month"] = (
    orders["order_purchase_timestamp"].dt.to_period("M")
)

first_purchase = (
    orders.groupby("customer_unique_id")["purchase_month"]
    .min()
    .rename("first_purchase_month")
)

orders = orders.merge(
    first_purchase,
    on="customer_unique_id"
)

orders["customer_type"] = (
    orders["purchase_month"] == orders["first_purchase_month"]
).map({
    True: "New",
    False: "Returning"
})

monthly_customers = pd.pivot_table(
    orders,
    index="purchase_month",
    columns="customer_type",
    values="customer_unique_id",
    aggfunc="nunique",
    fill_value=0
)

monthly_customers.tail()
```

```python
monthly_customers.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 6)
)

plt.title("Новые и повторные покупатели по месяцам")
plt.xlabel("Месяц")
plt.ylabel("Количество покупателей")
plt.xticks(rotation=45)
plt.legend(title="Тип покупателя")
plt.show()
```

### Анализ повторных покупок: задержка доставки и оценка первого заказа

```python
valid_orders = orders[
    ~orders["order_status"].isin(["canceled", "unavailable"])
].copy()

valid_orders = valid_orders.sort_values(
    ["order_purchase_timestamp", "order_id"]
)

first_orders = valid_orders.drop_duplicates(
    "customer_unique_id",
    keep="first"
).copy()

customer_stats = valid_orders.groupby("customer_unique_id").agg(
    orders_count=("order_id", "nunique")
)

first_orders = first_orders.merge(
    customer_stats,
    on="customer_unique_id"
)

first_orders["is_repeat"] = (
    first_orders["orders_count"] > 1
)

first_orders["delivery_status"] = np.where(
    first_orders["days_late"] > 0,
    "Delayed",
    "On time"
)

first_orders["review_group"] = pd.cut(
    first_orders["review_score"],
    bins=[0, 2, 5],
    labels=["Low (1-2)", "High (3-5)"]
)

repeat_analysis = first_orders.groupby(
    ["delivery_status", "review_group"],
    observed=False
).agg(
    customers_count=("customer_unique_id", "nunique"),
    returning_customers=("is_repeat", "sum"),
    repeat_rate=("is_repeat", "mean")
).reset_index()

repeat_analysis["repeat_rate"] = (
    repeat_analysis["repeat_rate"] * 100
).round(2)
```

```python

delivery_comparison = (
    first_orders.dropna(subset=["days_late"])
    .groupby(first_orders["days_late"] > 0)["is_repeat"]
    .mean() * 100
)

review_comparison = (
    first_orders.dropna(subset=["review_score"])
    .groupby(first_orders["review_score"] <= 2)["is_repeat"]
    .mean() * 100
)

if len(delivery_comparison) == 2:
    print(
        f"Доля повторных покупателей среди заказов с задержкой: "
        f"{delivery_comparison.loc[True]:.2f}%"
    )
    print(
        f"Доля повторных покупателей среди заказов без задержки: "
        f"{delivery_comparison.loc[False]:.2f}%"
    )

if len(review_comparison) == 2:
    print(
        f"Доля повторных покупателей при оценке 1–2: "
        f"{review_comparison.loc[True]:.2f}%"
    )
    print(
        f"Доля повторных покупателей при оценке 3–5: "
        f"{review_comparison.loc[False]:.2f}%"
    )
```
